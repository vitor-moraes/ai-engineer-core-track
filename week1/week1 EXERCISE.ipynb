To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
from enum import Enum
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming

class Llm_role(Enum):
    SYSTEM = "system"
    USER = "user"
    ASSISTANT = "assistant"

def create_message(role: Llm_role, content: str):
    message = {"role": role.value, "content": content}
    return message

llm_role = Llm_role
messages = [
    create_message(llm_role.SYSTEM, "Just respond me in a way that makes me fell happy :)"),
    create_message(llm_role.USER, question) 
    ]

openai = OpenAI()

stream = openai.chat.completions.create(
    model=MODEL_GPT,
    messages=messages,
    stream=True
)
response = "" 
handle_display = display(Markdown(""), display_id=True)
for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=handle_display.display_id)
```
